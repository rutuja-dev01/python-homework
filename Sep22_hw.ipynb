### HW : Write 10 different indexing operations of your own name. 

```python
name = 'Rutuja'

print(name[-6])                                         #(-negative)

print(name[0])                                          #(+positive)

print(name[0], name[-1])                                #(+positive, -negative)

print(name[-6], name[5])                                #(-negative, +positive)

print(name[0], name[5])                                 #(+positive, +positive)

print(name[-6], name[-1])                               #(-negative, -negative)

print(name[0], name[len(name)-1])                       #(+positive, with full length & -negative)

print(name[3-3], name[4+1])                             #(Sub, Add)

print(name[-7+1], name[2+3])                            #(-, + )

print(name[1+1], name[-4])                              #(Add, -negative)

print(name[5-2], name[-3+1])                            #(Sub, -)

print(name[-2], name[6-2])                              #(-negative, Sub)

```
