##### Que : WAP to check whether user given string is a palindrome or not.

```python
var = input("Enter a string : ")
rev = var[::-1]

if var == rev:
    print("It is Palindrome")
else: 
    print("It is not Palidrome")
```

##### Que : WAP to count the consonants in "Python".

```python
var = "Python"
count = 0

for ch in var: 
    if ch != 'a' and ch != 'e' and ch != 'i' and ch != 'o' and ch != 'u': 
        count = count + 1 

print(count)
```

##### Que : WAP to remove all spaces from " Python is easy".

```python
var = "Python is easy language"
result = ""

for ch in var:
    if ch != " ":
        result = result + ch

print(result)
```

##### Que : WAP to remove duplicate characters from "programming".

```python
var = "programming"
result = ""

for ch in var: 
    if ch not in result:
        result = result + ch

print(result)
```
